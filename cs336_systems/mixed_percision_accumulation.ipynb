```python
import torch.nn as nn

class ToyModel(nn.Module):
    def __init__(self, in_features: int, out_features: int):
        super().__init__()
        self.fc1 = nn.Linear(in_features, 10, bias=False)
        self.ln = nn.LayerNorm(10)
        self.fc2 = nn.Linear(10, out_features, bias=False)
        self.relu = nn.ReLU()

    def forward(self, x):
        print("Input dtype:", x.dtype)
        x = self.fc1(x)
        print("After FC1:", x.dtype)

        x = self.relu(x)
        print("After RELU:", x.dtype)
        
        x = self.ln(x)
        print("After norm:", x.dtype)
        
        x = self.fc2(x)
        print("After FC2:", x.dtype)

        return x
```

```python
import torch 
d_model = 128
d_ff = 1024

model = ToyModel(d_model, d_ff)

dtype = torch.bfloat16
device="cuda"

model = ToyModel(d_model, d_ff).to(device=device, dtype=torch.float32)

x = torch.randn(32, d_model, device=device)

with torch.autocast(device_type=device, dtype=torch.bfloat16):
    y = model(x)

print("Parameter dtype:", next(model.parameters()).dtype)
print("Output dtype:", y.dtype)
```
